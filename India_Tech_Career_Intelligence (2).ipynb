# **Outlier Detection and Analysis**

Outlier detection is performed to identify unusually high or low values in numerical columns.
The identified values are examined to determine whether they are genuine observations or potential data errors before deciding whether any treatment is required.

- **Selecting Numerical Columns for Outlier Analysis**

The important numerical columns related to salary, stipend, skills, and applicants are selected for outlier analysis.

```python
# Select numerical columns for outlier checking

num_cols = ["salary_minimum","salary_maximum","stipend_amount","skill_count","applicants"]


```

- **Checking Numerical Data**

```python
df[num_cols].describe().T
```

- **Calculating Quartiles**

The first quartile (Q1) represents the 25th percentile and the third quartile (Q3) represents the 75th percentile.
These values are used to calculate the Interquartile Range (IQR) for identifying potential outliers.

```python
# Calculate Q1 and Q3

Q1 = df["salary_maximum"].quantile(0.25)
Q3 = df["salary_maximum"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

- **Calculating Interquartile Range (IQR)**

The Interquartile Range (IQR) represents the middle 50% of the salary values.
It is calculated as the difference between Q3 and Q1.

```python
# Calculate IQR

IQR = Q3 - Q1

print("IQR:", IQR)
```

- **Calculating Outlier Boundaries**

The lower and upper boundaries are calculated using the IQR method.
Values below the lower boundary or above the upper boundary are considered potential outliers.

```python
# Calculate lower and upper bounds

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

- **Identifying Potential Salary Outliers**

Salary values outside the calculated lower and upper boundaries are identified as potential outliers.
These values are not removed automatically because they may represent genuine high-paying technology jobs.

```python
# Identify potential outliers

potential_outliers = df[
    (df["salary_maximum"] < lower_bound) |
    (df["salary_maximum"] > upper_bound)]

print("Number of potential outliers:", len(potential_outliers))
```

The IQR method was used to identify potential outliers in `salary_maximum`. No potential outliers were detected, so no outlier treatment was required for this column.

- **Checking Salary Minimum for Outliers**

```python
Q1_min = df["salary_minimum"].quantile(0.25)
Q3_min = df["salary_minimum"].quantile(0.75)

IQR_min = Q3_min - Q1_min

lower_min = Q1_min - 1.5 * IQR_min
upper_min = Q3_min + 1.5 * IQR_min

print("Q1:", Q1_min)
print("Q3:", Q3_min)
print("IQR:", IQR_min)
print("Lower Bound:", lower_min)
print("Upper Bound:", upper_min)
```

```python
potential_outliers_min = df[
    (df["salary_minimum"] < lower_min) |
    (df["salary_minimum"] > upper_min)]

print("Potential salary minimum outliers:", len(potential_outliers_min))
```

Potential outliers in salary values were identified using the IQR method.
The identified observations were reviewed instead of being automatically removed, as unusually high salaries may represent genuine technology job opportunities.
Only clearly incorrect values would be treated if identified.

# **Feature Transformations**

Feature transformations are performed to create useful derived features from existing columns.
These new features help analyze salary trends, application periods, and hiring trends over time.

- **Creating Salary Midpoint**

The salary midpoint is calculated using the minimum and maximum salary values.
It provides a single representative salary value that can be used for salary comparisons and visualizations.

```python
# Calculate salary midpoint

df["salary_midpoint"] = (df["salary_minimum"] + df["salary_maximum"]) / 2
```

```python
df[["salary_minimum","salary_maximum","salary_midpoint"]].head()
```

- **Creating Salary Range**

The salary range represents the difference between the maximum and minimum salary offered for a job.
It helps identify how widely the salary can vary within a job posting.

```python
# Calculate salary range

df["salary_range"] = (df["salary_maximum"] - df["salary_minimum"])
```

- **Creating Salary Range Percentage**

The salary range percentage measures the salary variation relative to the minimum salary. It helps compare salary flexibility across job postings with different salary levels.

```python
df["salary_range_percent"] = (df["salary_range"] / df["salary_minimum"]) * 100
```

- **Extracting Posting Month**

The month is extracted from the job posting date to analyze monthly hiring patterns.

```python
# Extract posting month

df["posted_month"] = df["posted_date"].dt.month
```

- **Creating Posting Month Name**

The month name is extracted from the posting date to make time-based analysis and visualizations easier to understand.

```python
# Extract posting month name

df["posted_month_name"] = (df["posted_date"].dt.month_name())
```

- **Creating Posting Quarter**

The quarter is extracted from the job posting date to analyze hiring activity across different quarters of the year.

```python
# Extract posting quarter

df["posted_quarter"] = (df["posted_date"].dt.quarter)
```

### Creating Individual Skill Records

```python
# Convert each skill into a separate row

skills_df = skills_df.explode("skills_extracted")

skills_df.head(10)
```

### 1. Salary and Career Growth
The average salary midpoint is approximately ₹849.95K, while average career
growth is 81.51%. The correlation between salary and career growth is almost
zero, indicating that higher salary does not necessarily correspond to higher
career growth in this dataset.

### 2. Experience Level
Mid-Senior Level jobs represent approximately 59.9% of the analyzed postings,
while Freshers / Entry Level jobs represent approximately 40.1%.

### 3. Work Mode
Remote and Hybrid jobs together represent approximately 79.7% of the
analyzed postings, showing strong representation of flexible work arrangements.

### 4. Technology Skills
Python and Linux are the most frequently represented skills in the dataset,
with more than 33K job postings each.

### 5. Hiring Trend
The monthly job-posting trend shows a sharp decline in February followed by
a strong recovery in March.
